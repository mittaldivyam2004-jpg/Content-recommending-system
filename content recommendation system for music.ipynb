```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

### Data Loading and Exploration

For demonstration, a small dataset with the same structure
as the Firestore database is used.

```python
songs = pd.DataFrame({
    "song_id": [1, 2, 3, 4],
    "song_name": ["Tum Hi Ho", "Believer", "Perfect", "Senorita"],
    "artist": ["Arijit Singh", "Imagine Dragons", "Ed Sheeran", "Shawn Mendes"],
    "genre": ["Romantic", "Rock", "Romantic", "Pop"],
    "mood": ["Sad", "Energetic", "Romantic", "Happy"]
})

songs
```

### Data Cleaning & Feature Engineering

Song attributes such as genre, mood, and artist are combined
to create a single feature.


```python
songs["features"] = songs["genre"] + " " + songs["mood"] + " " + songs["artist"]
songs[["song_name", "features"]]
```

## Model / System Design

### AI Technique Used
- Hybrid Recommendation System
- Content-Based Filtering
- Collaborative Filtering

### System Architecture
Frontend → Backend → Firebase


```python
vectorizer = TfidfVectorizer(
    max_features=20,
    stop_words="english"
)

tfidf_matrix = vectorizer.fit_transform(songs["features"])
similarity_matrix = cosine_similarity(tfidf_matrix)

similarity_matrix
```

## Core Implementation

The recommendation pipeline compares song features
and suggests the most similar songs.


```python
def recommend_song(song_index, top_n=2):
    scores = list(enumerate(similarity_matrix[song_index]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    indices = [i[0] for i in scores[1:top_n+1]]
    return songs.iloc[indices][["song_name", "artist", "genre"]]

recommend_song(0)
```
